### ทำนายว่า Cholesterol Checked ไม่ได้มีค่าเท่ากับ Told High Cholesterol แต่ถ้ามันมีค่าเท่ากันหนึ่งต่อหนึ่งก็อาจจะปล่อยให้ Told High Cholesterol เป็นค่าว่างต่อไปได้เลย เพราะเกี่ยวเนื่องกับการที่ ว่าง เพราะ ยังไม่ได้ตรวจ

```python
train_not_null = train_df[train_df["Told High Cholesterol"].notnull()]
print("Cholesterol Checked from Not null Told High Cholesterol : ", train_not_null["Cholesterol Checked"].value_counts(normalize=False))
print("Cholesterol Checked from Not null Told High Cholesterol (Normalized) : ", train_not_null["Cholesterol Checked"].value_counts(normalize=True))
print("Cholesterol Checked from No Told High Cholesterol : ",train_df[train_df["Told High Cholesterol"] == "No"]["Cholesterol Checked"].value_counts(normalize=False))
print("Cholesterol Checked from No Told High Cholesterol (Normalized) : ",train_df[train_df["Told High Cholesterol"] == "No"]["Cholesterol Checked"].value_counts(normalize=True))
print("Cholesterol Checked from Yes Told High Cholesterol : ",train_df[train_df["Told High Cholesterol"] == "Yes"]["Cholesterol Checked"].value_counts(normalize=False))
print("Cholesterol Checked from Yes Told High Cholesterol (Normalized): ",train_df[train_df["Told High Cholesterol"] == "Yes"]["Cholesterol Checked"].value_counts(normalize=True))
```
